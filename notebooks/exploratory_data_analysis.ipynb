# Personal Expense Tracker - Exploratory Data Analysis (EDA)
This notebook demonstrates exploratory data analysis on personal financial transactions, category aggregations, monthly trends, and visualization techniques.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style='whitegrid')
df = pd.read_csv('../data/expenses_sample.csv')
df.head()
```

```python
# Data Overview & Summary Statistics
print('Dataset Shape:', df.shape)
print('Missing Values:\n', df.isnull().sum())
df.describe()
```

```python
# Category-Wise Spending Breakdown
expenses = df[df['Category'] != 'Income']
cat_spend = expenses.groupby('Category')['Amount'].sum().sort_values(ascending=False)
print('Category Totals (INR):\n', cat_spend)

plt.figure(figsize=(10, 5))
sns.barplot(x=cat_spend.values, y=cat_spend.index, palette='mako')
plt.title('Spending by Category')
plt.xlabel('Amount (INR)')
plt.show()
```

```python
# Payment Method Share
pm_spend = expenses.groupby('Payment_Method')['Amount'].sum()
plt.figure(figsize=(6, 6))
plt.pie(pm_spend.values, labels=pm_spend.index, autopct='%1.1f%%', startangle=140)
plt.title('Payment Method Distribution')
plt.show()
```
